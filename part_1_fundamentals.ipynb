### 10.3

````python

    # Your code here
    
def validate_username(username):
    if len(username) < 3 or len(username)> 20: # The username needs to be between 3 to 20 carracters
        return False
    if not username[0].isalphs(): # Only letters are allowed and the first caracter needs to be a letter
        return(False)
    if not username.replace("_", "").isalnum(): # All signs except _ is allowed
        return False
    
    return True


print(validate_username("dkii4459anns4fd#")) # prints false
print(validate_username("dkii4459anns4fd")) # prints true
````

### Task 16
Write a function `describe_temperature(temp)` that returns:

- `"warm"` if it is 25 or above
Use `if`, `elif`, and `else`.
Call the function a few times to show that it works.

### Task 22

You have a list consisting of dictionaries. Each dictionary represents data for an individual student in a system.

You want to create a new list that only consists of students where "Passed" is True, how do you do it?

**Tip** Utilize lists, loops and if-statements.

**Answer with code**


````python


# Lägg till
# Example list of students 
students = [
    {"name": "Anna", "grade": "G"},
    {"name": "Bob", "grade": "IG"},
    {"name": "Clara", "grade": "G"},
    {"name": "David", "grade": "IG"},
    {"name": "Emma", "grade": "G"}
]

# Two empty lists for the result
passed = []
failed = []

# Loop through every student
for student in students:
    if student["grade"] == "G":          # Check if they passed (G = Passed)
        passed.append(student)
    else:
        failed.append(student)           # Everything else (IG, etc.) goes to failed

# Show the results
print("Passed students:")
print(passed)

print("\nFailed students:")
print(failed)
````
